# Extracción y Almacenamiento Seguro de Datos vía API

**Objetivo:** desarrollar un script en Python que realice solicitudes GET a una API pública, extraiga información, aplique una transformación básica y almacene los resultados en formato JSON.

**API utilizada:** JSONPlaceholder.

**Herramientas:** Python, Requests, Python-dotenv y Google Colab.

**Resultado esperado:** obtener información de usuarios desde una API externa y guardarla en un archivo JSON.

```python
# Importación de librerías

import requests
import json
import os

from dotenv import load_dotenv
```

```python
# Configuración de la API

API_URL = "https://jsonplaceholder.typicode.com/users"

print("API configurada correctamente")
```

```python
# Primera conexión con la API

try:
    response = requests.get(API_URL, timeout=10)

    response.raise_for_status()

    usuarios = response.json()

    print("Conexión exitosa con la API")
    print(f"Cantidad de usuarios obtenidos: {len(usuarios)}")

except requests.exceptions.RequestException as e:
    print(f"Error al consultar la API: {e}")
```

```python
# Transformación y limpieza de datos

usuarios_limpios = []

for usuario in usuarios:
    usuario_filtrado = {
        "id": usuario["id"],
        "nombre": usuario["name"],
        "usuario": usuario["username"],
        "email": usuario["email"],
        "ciudad": usuario["address"]["city"],
        "empresa": usuario["company"]["name"]
    }

    usuarios_limpios.append(usuario_filtrado)

print("Transformación realizada correctamente")
print(f"Registros procesados: {len(usuarios_limpios)}")
```

```python
# Visualización de los datos transformados

print(json.dumps(usuarios_limpios[0], indent=4, ensure_ascii=False))
```

```python
# Almacenamiento de los datos en formato JSON

nombre_archivo = "data_extracted.json"

with open(nombre_archivo, "w", encoding="utf-8") as archivo:
    json.dump(usuarios_limpios, archivo, indent=4, ensure_ascii=False)

print(f"Datos guardados correctamente en {nombre_archivo}")
```

```python
# Verificación del archivo generado

if os.path.exists(nombre_archivo):
    print("El archivo existe correctamente")
    print(f"Tamaño del archivo: {os.path.getsize(nombre_archivo)} bytes")
else:
    print("No se encontró el archivo")
```

```python

import requests
import json
import os
from dotenv import load_dotenv

# 1. Cargar variables de entorno
load_dotenv()

API_URL = os.getenv("API_URL")

# 2. Verificar configuración
if not API_URL:
    raise ValueError("No se encontró la variable API_URL en el archivo .env")

# 3. Extracción de datos
try:
    response = requests.get(API_URL, timeout=10)
    response.raise_for_status()

    usuarios = response.json()

    if not isinstance(usuarios, list):
        raise ValueError("La respuesta de la API no tiene el formato esperado")

    # 4. Transformación de datos
    usuarios_limpios = []

    for usuario in usuarios:
        usuario_filtrado = {
            "id": usuario["id"],
            "nombre": usuario["name"],
            "usuario": usuario["username"],
            "email": usuario["email"],
            "ciudad": usuario["address"]["city"],
            "empresa": usuario["company"]["name"]
        }

        usuarios_limpios.append(usuario_filtrado)

    # 5. Almacenamiento en JSON
    os.makedirs("data", exist_ok=True)

    nombre_archivo = "data/data_extracted.json"

    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        json.dump(
            usuarios_limpios,
            archivo,
            indent=4,
            ensure_ascii=False
        )

    print("Extracción realizada correctamente")
    print(f"Usuarios obtenidos: {len(usuarios_limpios)}")
    print(f"Archivo generado: {nombre_archivo}")

# 6. Manejo de errores
except requests.exceptions.HTTPError as e:
    print(f"Error HTTP: {e}")

except requests.exceptions.Timeout:
    print("Error: se agotó el tiempo de espera de la API")

except requests.exceptions.ConnectionError:
    print("Error: no se pudo establecer conexión con la API")

except requests.exceptions.RequestException as e:
    print(f"Error en la petición: {e}")

except (ValueError, KeyError) as e:
    print(f"Error al procesar los datos: {e}")
```

```python

%%writefile extract_data.py

import requests
import json
import os
from dotenv import load_dotenv

# 1. Cargar variables de entorno
load_dotenv()

API_URL = os.getenv("API_URL")

# 2. Verificar configuración
if not API_URL:
    raise ValueError("No se encontro API_URL en el archivo .env")

# 3. Extraccion de datos
try:
    response = requests.get(API_URL, timeout=10)
    response.raise_for_status()

    usuarios = response.json()

    if not isinstance(usuarios, list):
        raise ValueError("Formato de respuesta inesperado")

    # 4. Transformacion de datos
    usuarios_limpios = []

    for usuario in usuarios:
        usuario_filtrado = {
            "id": usuario["id"],
            "nombre": usuario["name"],
            "usuario": usuario["username"],
            "email": usuario["email"],
            "ciudad": usuario["address"]["city"],
            "empresa": usuario["company"]["name"]
        }

        usuarios_limpios.append(usuario_filtrado)

    # 5. Almacenamiento
    os.makedirs("data", exist_ok=True)

    with open("data/data_extracted.json", "w", encoding="utf-8") as archivo:
        json.dump(usuarios_limpios, archivo, indent=4, ensure_ascii=False)

    print("Extraccion realizada correctamente")
    print(f"Usuarios obtenidos: {len(usuarios_limpios)}")
    print("Archivo generado: data/data_extracted.json")

except requests.exceptions.HTTPError as e:
    print(f"Error HTTP: {e}")

except requests.exceptions.Timeout:
    print("Error: tiempo de espera agotado")

except requests.exceptions.ConnectionError:
    print("Error de conexion con la API")

except requests.exceptions.RequestException as e:
    print(f"Error en la peticion: {e}")

except (ValueError, KeyError) as e:
    print(f"Error al procesar los datos: {e}")
```
